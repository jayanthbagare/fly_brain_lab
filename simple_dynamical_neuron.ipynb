$$
\begin{align}

Where $\tau$ = Time Constant, how fast the voltage forgets and relaxes back.  
$V_{th}$ is the Threshold voltage where the neuron fires a spike when it reaches it.  
$V_{rest}$ is the resting voltage.  
$V_{reset}$ is the reset voltage, where when the firing neuron drops to from the voltage threshold.  
R is the resistance.  
I is the incoming current.
